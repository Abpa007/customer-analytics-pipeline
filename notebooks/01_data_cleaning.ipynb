```python
import pandas as pd

# Load the raw dataset
df = pd.read_csv('../data/customers.csv')

# View the first 5 rows
print("--- First 5 Rows ---")
print(df.head())

# Check dataset structure and data types
print("\n--- Dataset Info ---")
print(df.info())
```

```python
# Check for null values in each column
print("--- Missing Values ---")
print(df.isnull().sum())

# Inspect column names for trailing spaces or casing issues
print("\n--- Original Column Names ---")
print(df.columns.tolist())
```

```python
# Standardize column names (lowercase, replace spaces with underscores)
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')

# Drop completely empty rows and fill missing values
df = df.dropna(how='all')
df.fillna('Unknown', inplace=True)

# Verify cleaned columns and shape
print("--- Cleaned Columns ---")
print(df.columns.tolist())
print(f"Final Shape: {df.shape}")

# Save to cleaned dataset
df.to_csv('../data/cleaned_customers.csv', index=False)
print("\n✅ Successfully saved cleaned dataset!")
```
